# Sleep and Academic Perfomance
## 1. Dataset Setup

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("../data/cmu-sleep.csv")

data.head(10)
data.shape
data.columns
data.info()
```

## 2. Exploratory Data Analysis
### 2.1 Summary Statitstics

```python
data[["cum_gpa","daytime_sleep","TotalSleepTime"]].describe()
```

### 2.2 Distributions

```python
plt.hist(data["TotalSleepTime"], bins=100)
plt.xlabel("Sleep time in minutes")
plt.ylabel("Students")
print("Mean for term gpa = ",data["term_gpa"].mean())
print("Median for term gpa = ",data["term_gpa"].median())
print("Difference between median and mean in term gpa = ", abs(data["term_gpa"].mean() - data["term_gpa"].median()) )
```

### 2.3 Relationships Between Variables

```python
plt.scatter(data["cum_gpa"],data["term_gpa"])
plt.xlabel("Cumulative GPA Data")
plt.ylabel("Term GPA Data")
print("r =" ,data["cum_gpa"].corr(data["term_gpa"]))
```
